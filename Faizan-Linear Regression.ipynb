```python
#  Here  X implies the independent feature
x = df.drop('MEDV', axis=1) # All column except 'MEDV'
# Here Y implies the dependent feature on target
y = df['MEDV']
```

```python
# Scale the feature 
scale = StandardScaler()
#We will pass X values inside StandardScaler to scale the values
# to scale the value
x_scaled = scale.fit_transform(x)
```

```python
# split the data into training (80%) and testing (20%)
x_train, x_test, y_train, y_test = train_test_split(
    x_scaled , y,
    test_size=0.20, random_state=42)
```

```python
x_train.shape
```

```python
# Initialize the linear regression model
model = LinearRegression()

#Fit the model on the training data
model.fit(x_train, y_train)
```

```python
# predict Y values on the test set's X values
y_pred = model.predict(x_test)
y_pred
```

```python
# calculate mean squared error
mse = mean_squared_error(y_test, y_pred)

# calculate root mean squared error (RMSE)
rmse = np.sqrt(mse)

# calculate r-squared value (accuracy)
r2 = r2_score(y_test, y_pred)
```
